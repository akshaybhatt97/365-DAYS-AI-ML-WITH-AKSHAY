# Day 86/365 — Backpropagation

Yesterday we learned that the loss function tells a neural network how wrong its prediction is.

Today we answer the next question:

**How does the network use that error to improve its weights?**

That process is called **Backpropagation**.


```python
import numpy as np
```

## 1. The training loop

1. Forward propagation
2. Compute loss
3. Backpropagate gradients
4. Update weights
5. Repeat


## 2. Tiny example

Let:

`x = 2`

`w = 3`

Target:

`y = 10`

Prediction:

`ŷ = wx = 6`


```python
x = 2.0
w = 3.0
y = 10.0

y_hat = w * x
print("Prediction:", y_hat)
```

## 3. Compute loss

Use squared error:

`L = (y - ŷ)²`

So:

`L = (10 - 6)² = 16`


```python
loss = (y - y_hat)**2
print("Loss:", loss)
```

## 4. Compute the gradient

We want:

`dL/dw`

Using the chain rule:

`dL/dw = dL/dŷ × dŷ/dw`

For squared error:

`dL/dŷ = 2(ŷ - y)`

and:

`dŷ/dw = x`

Therefore:

`dL/dw = 2(ŷ - y)x`


```python
gradient = 2 * (y_hat - y) * x
print("Gradient dL/dw:", gradient)
```

Here the gradient is negative.

That means increasing the weight will reduce the loss.


## 5. Update the weight

Gradient descent update:

`w_new = w - learning_rate × gradient`

Let:

`learning_rate = 0.1`


```python
learning_rate = 0.1

new_w = w - learning_rate * gradient

print("Old weight:", w)
print("New weight:", new_w)
```

## 6. Check whether the loss improved


```python
new_prediction = new_w * x
new_loss = (y - new_prediction)**2

print("New prediction:", new_prediction)
print("New loss:", new_loss)
```

The loss becomes smaller.

That is the core idea behind learning:

**change weights in the direction that reduces loss.**


## 7. Why is it called backpropagation?

Forward:

`Input → Layers → Prediction → Loss`

Backward:

`Loss → Gradients → Earlier Layers`

The error signal is propagated backward through the network.

Each derivative tells us how much a parameter contributed to the final loss.


## 8. Chain rule

Neural networks are compositions of many functions.

For a path like:

`w → z → activation → prediction → loss`

the chain rule lets us connect local derivatives to calculate:

`dLoss/dw`

This is the mathematical engine behind backpropagation.


## 9. Gradient vs learning rate

**Gradient**

Tells us the direction and sensitivity of the loss.

**Learning rate**

Controls how large a step we take.

Update:

`weight = weight - learning_rate × gradient`


## 10. Why it matters

Neural networks can contain millions or billions of parameters.

Backpropagation efficiently computes gradients for all of them.

Frameworks such as PyTorch and TensorFlow automate this using automatic differentiation.


## Simple takeaway

Forward propagation:

`Weights → Prediction → Loss`

Backpropagation:

`Loss → Gradients → Weight Updates`

> **Forward propagation tells the network what it predicted. Backpropagation tells it how to change so the next prediction can be better.**
