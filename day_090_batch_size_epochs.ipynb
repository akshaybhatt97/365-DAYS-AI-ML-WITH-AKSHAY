# Day 90/365 — Batch Size, Epochs & Mini-Batch Training

After learning optimizers, the next practical question is:

**How much data should the model process before each weight update?**

That brings us to:

- Epoch
- Batch
- Batch Size
- Iteration / Step
- Mini-batch training


```python
import numpy as np
import matplotlib.pyplot as plt
```

## 1. Epoch

An **epoch** means the model has seen the entire training dataset once.

If your training set has `1,000 samples`, then after one epoch all 1,000 samples have been used once.


## 2. Batch

A **batch** is a smaller subset of the training data.

If:

`Dataset size = 1,000`

and:

`Batch size = 100`

then:

`1,000 / 100 = 10 batches`


## 3. Iteration / Step

One batch normally produces:

1. forward pass
2. loss calculation
3. backpropagation
4. optimizer update

That is one training **step** or **iteration**.


```python
dataset_size = 1000
batch_size = 100

steps_per_epoch = dataset_size // batch_size

print("Steps per epoch:", steps_per_epoch)
```

## 4. Example

Dataset = `1,000`

Batch size = `100`

Epochs = `5`

Then:

`10 batches per epoch`

and:

`5 × 10 = 50 optimizer updates`


```python
epochs = 5
total_updates = epochs * steps_per_epoch
print("Total optimizer updates:", total_updates)
```

## 5. Full-batch vs Stochastic vs Mini-batch

### Full-batch Gradient Descent
Use the entire dataset before one update.

### Stochastic Gradient Descent
Use one sample before each update.

### Mini-batch Gradient Descent
Use a small batch such as `32`, `64`, or `128` before each update.

Mini-batch training is the standard approach in modern deep learning.


## 6. Why mini-batches work well

They balance:

- efficient GPU computation
- frequent weight updates
- manageable memory use
- less noisy gradients than one-sample SGD


## 7. Batch-size trade-offs

### Smaller batch
- lower memory usage
- more frequent updates
- noisier gradient estimates

### Larger batch
- smoother gradient estimates
- better parallel hardware utilization
- higher memory usage
- fewer updates per epoch


```python
batch_sizes = [16,32,64,128,256]
dataset_size = 1024

for b in batch_sizes:
    steps = int(np.ceil(dataset_size / b))
    print(f"Batch size {b:3d} -> {steps:2d} steps per epoch")
```

## 8. Useful formula

If:

`N = number of samples`

and:

`B = batch size`

then:

`steps per epoch = ceil(N / B)`


## 9. Batch size and learning rate interact

When batch size changes significantly, the best learning rate may also change.

So batch size should not be tuned completely in isolation.


## Simple takeaway

- **Epoch** = one full pass through the dataset
- **Batch** = one subset of the dataset
- **Batch Size** = samples in that subset
- **Step / Iteration** = one optimizer update
- **Mini-batch training** = the practical middle ground used in most deep-learning systems

> **Batch size controls how much data the model uses before each learning step.**
