# Day 92/365 — Dropout

Yesterday we learned **Batch Normalization**.

Today we look at a popular regularization technique:

**Dropout**

During training, Dropout randomly switches off some neurons.

The goal is to reduce over-reliance on particular neurons and improve generalization.


```python
import numpy as np
```

## 1. Tiny example

Suppose a hidden layer outputs:

`[0.8, 0.4, 0.9, 0.2]`

With dropout, some activations are randomly set to zero.


```python
rng = np.random.default_rng(42)

activations = np.array([0.8, 0.4, 0.9, 0.2])

dropout_rate = 0.5
keep_prob = 1 - dropout_rate

mask = rng.binomial(
    n=1,
    p=keep_prob,
    size=activations.shape
)

dropped = activations * mask

print("Original:", activations)
print("Mask:", mask)
print("After dropout:", dropped)
```

## 2. Inverted dropout

Modern frameworks usually scale activations during training:

`output = activation × mask / keep_probability`

This keeps the expected activation scale roughly unchanged.


```python
scaled = activations * mask / keep_prob
print("Scaled dropout output:", scaled)
```

## 3. Training vs inference

### During training
Dropout is ON.

Different neurons may be disabled on each training step.

### During inference
Dropout is OFF.

The full network is used.


## 4. Why can Dropout help?

Without regularization, some neurons may become too dependent on one another.

Dropout makes the network learn representations that remain useful even when some units are missing.

This can reduce overfitting and improve generalization.


## 5. Typical dropout rates

Values such as:

`0.1`

`0.2`

`0.5`

are common starting points, depending on the architecture.

Too much dropout can make learning unnecessarily difficult.


## 6. Dropout in modern architectures

Dropout is used differently across architectures.

- CNNs may apply it to selected layers.
- Transformers often use dropout around attention or feed-forward blocks.
- Other regularization methods may also be used.

So Dropout is a tool, not a universal rule.


## Simple takeaway

Dropout randomly removes some neurons during training so the network cannot depend too much on any one path.

> **The network becomes more robust by learning while some of its neurons are temporarily unavailable.**
